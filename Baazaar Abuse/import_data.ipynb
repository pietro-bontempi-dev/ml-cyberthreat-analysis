```python
import numpy as np
import pandas as pd
from sqlalchemy import create_engine, text

# 1. Configurações de Conexão
DB_USER = "root"  # Seu usuário
DB_PASS = "25644669Pe#!"  # Sua senha
DB_HOST = "localhost"
DB_PORT = "3306"
DB_NAME = "bazaar_malware_db"

DATABASE_URL = f"mysql+pymysql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}?charset=utf8mb4"
engine = create_engine(DATABASE_URL)

# 2. Leitura e Limpeza do CSV
importacao = input("Digite o nome do arquivo que você irá importar: ") + ".csv"

colunas = [
    "first_seen_utc",
    "sha256_hash",
    "md5_hash",
    "sha1_hash",
    "reporter",
    "file_name",
    "file_type_guess",
    "mime_type",
    "signature",
    "clamav",
    "vtpercent",
    "imphash",
    "ssdeep",
    "tlsh",
]

print("Lendo o arquivo CSV...")
df_bazaar_malware = pd.read_csv(
    importacao,
    comment="#",
    skipinitialspace=True,
    names=colunas,
    header=0,
    encoding="utf-8",
    on_bad_lines="skip",
    low_memory=False,
)

# Remover cabeçalhos acidentais
df_bazaar_malware = df_bazaar_malware[
    df_bazaar_malware["first_seen_utc"] != "first_seen_utc"
]

# Substituir 'n/a' por None
df_bazaar_malware.replace("n/a", None, inplace=True)

# Remover duplicatas de SHA256 no próprio DataFrame
df_bazaar_malware.drop_duplicates(subset=["sha256_hash"], inplace=True)

# 3. Povoando as Tabelas Auxiliares
print("Populando tabelas auxiliares...")

with engine.begin() as connection:
    # Repórteres
    reporters = (
        df_bazaar_malware[["reporter"]]
        .dropna()
        .drop_duplicates()
        .rename(columns={"reporter": "name"})
    )
    for name in reporters["name"]:
        connection.execute(
            text("INSERT IGNORE INTO reporters (name) VALUES (:name)"),
            {"name": name},
        )

    # Tipos de Arquivo
    file_types = df_bazaar_malware[
        ["file_type_guess", "mime_type"]
    ].drop_duplicates()
    for _, row in file_types.iterrows():
        connection.execute(
            text(
                "INSERT IGNORE INTO file_types (type_guess, mime_type) VALUES (:type_guess, :mime_type)"
            ),
            {
                "type_guess": row["file_type_guess"],
                "mime_type": row["mime_type"],
            },
        )

    # Assinaturas
    signatures = (
        df_bazaar_malware[["signature"]]
        .dropna()
        .drop_duplicates()
        .rename(columns={"signature": "signature_name"})
    )
    for name in signatures["signature_name"]:
        connection.execute(
            text("INSERT IGNORE INTO signatures (signature_name) VALUES (:name)"),
            {"name": name},
        )

    # ClamAV
    clamav = (
        df_bazaar_malware[["clamav"]]
        .dropna()
        .drop_duplicates()
        .rename(columns={"clamav": "clamav_name"})
    )
    for name in clamav["clamav_name"]:
        connection.execute(
            text(
                "INSERT IGNORE INTO clamav_signatures (clamav_name) VALUES (:name)"
            ),
            {"name": name},
        )

    # 4. Mapeamento de IDs
    print("Mapeando relacionamentos...")
    rep_df = pd.read_sql(
        "SELECT id AS reporter_id, name AS reporter FROM reporters", connection
    )
    ft_df = pd.read_sql(
        "SELECT id AS file_type_id, type_guess AS file_type_guess, mime_type FROM file_types",
        connection,
    )
    sig_df = pd.read_sql(
        "SELECT id AS signature_id, signature_name AS signature FROM signatures",
        connection,
    )
    clam_df = pd.read_sql(
        "SELECT id AS clamav_id, clamav_name AS clamav FROM clamav_signatures",
        connection,
    )

    df_final = df_bazaar_malware.merge(rep_df, on="reporter", how="left")
    df_final = df_final.merge(
        ft_df, on=["file_type_guess", "mime_type"], how="left"
    )
    df_final = df_final.merge(sig_df, on="signature", how="left")
    df_final = df_final.merge(clam_df, on="clamav", how="left")

    df_final["vtpercent"] = pd.to_numeric(
        df_final["vtpercent"], errors="coerce"
    )

    # Filtrar registros cujo sha256_hash JÁ EXISTE no MySQL
    print("Verificando hashes que já existem no banco de dados...")
    existing_hashes = pd.read_sql(
        "SELECT sha256_hash FROM malware_samples", connection
    )
    if not existing_hashes.empty:
        df_final = df_final[
            ~df_final["sha256_hash"].isin(existing_hashes["sha256_hash"])
        ]

    samples_df = df_final[
        [
            "first_seen_utc",
            "sha256_hash",
            "md5_hash",
            "sha1_hash",
            "file_name",
            "reporter_id",
            "file_type_id",
            "signature_id",
            "clamav_id",
            "vtpercent",
            "imphash",
            "ssdeep",
            "tlsh",
        ]
    ]

    print(f"Novas amostras a serem inseridas: {len(samples_df)}")

    # 5. Inserção Segura na Tabela Principal
    if len(samples_df) > 0:
        samples_df.to_sql(
            "malware_samples",
            connection,
            if_exists="append",
            index=False,
            chunksize=5000,
        )

print("Carga finalizada com sucesso! Amostras duplicadas foram ignoradas.")
```
